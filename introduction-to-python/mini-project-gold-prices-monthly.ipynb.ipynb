# Monthly Price Data – Mini Project

## Project goals

- Load a real CSV file (`monthly.csv`) using pure Python.
- Parse raw text lines into clean date and price columns without using pandas.
- Convert the data into NumPy arrays for further analysis.
- Build a 2D NumPy table with rows of the form `[year, month, price]`.
- Practice basic text processing in Python (`strip`, `split`, loops, lists) and NumPy (`array`, `dtype`, `shape`).

```python
import numpy as np
filename = "monthly.csv"
f = open(filename, "r")
lines = f.readlines()
f.close()
```

```python
data_lines = lines[1:]

dates = []

for line in data_lines:
    parts = line.strip().split(",")
    date_str = parts[0] 
    dates.append(date_str) 
```

```python

prices = []

for line in data_lines:
    parts = line.strip().split(",")
    price_str = parts[1]
    prices.append(price_str)
```

```python
np_dates = np.array(dates)
np_prices = np.array(prices)
print(np_dates)
print(np_prices)
```

```python
years = []
months = []

for d in np_dates:
    y_str, m_str = d.split("-")
    years.append(int(y_str))
    months.append(int(m_str))

np_years = np.array(years)
np_months = np.array(months)
```

```python
np_prices = np.array(prices, dtype= float)
```

```python
table = np.column_stack ((np_years, np_months, np_prices))
print(table.shape)
print(table[:5])
```
