# Monthly Gold Price Data – Mini Project

This mini project explores a historical time series of **monthly gold prices** from 1833 to 2026 stored in a CSV file (`monthly.csv`).  
The goal is to practice working with raw text files in pure Python and NumPy instead of using high‑level libraries like pandas.

We will:

- load the CSV file manually using basic file I/O,
- parse raw text lines into clean `date` and `price` columns,
- convert these columns into NumPy arrays,
- split the `YYYY-MM` date strings into separate `year` and `month` integers,
- build a 2D NumPy table with rows of the form `[year, month, price]`,
- briefly inspect the resulting structure and discuss what it represents.

```python
import numpy as np
filename = "monthly.csv"
f = open(filename, "r")
lines = f.readlines()
f.close()
```

## 1. Loading the raw CSV file

The CSV file has the following structure:

- first line: header with column names, e.g. `Date,Price`,
- all remaining lines: monthly observations in the format `YYYY-MM,VALUE`.

Instead of using pandas, we use **plain Python file I/O** to make the parsing steps explicit:

1. open the file in read mode,
2. read all lines into a list of strings,
3. close the file handle.

```python
data_lines = lines[1:]

dates = []

for line in data_lines:
    parts = line.strip().split(",")
    date_str = parts[0] 
    dates.append(date_str)
```

## 2. Parsing the `Date` column

The `lines` list contains the header plus all data rows.  
We:

- skip the header (`lines[1:]`),
- strip each line to remove the newline character,
- split by comma to separate the date and the price,
- keep only the first element (`YYYY-MM`) and collect it in a `dates` list.

This is a good exercise in basic **string processing** (`strip`, `split`) and list handling in Python.

```python
prices = []

for line in data_lines:
    parts = line.strip().split(",")
    price_str = parts[1]
    prices.append(price_str)
```

## 3. Parsing the `Price` column

We repeat a similar procedure for the `Price` column:

- reuse `data_lines` without the header,
- split each line into `[date, price]`,
- take the second element (price) and append it to a `prices` list.

At this stage both `dates` and `prices` are plain Python lists containing strings.

```python
np_dates = np.array(dates)
np_prices = np.array(prices)
print(np_dates)
print(np_prices)
```

## 4. Converting lists to NumPy arrays

To enable numerical operations and vectorised processing, we convert both lists to NumPy arrays.

- `np_dates` is still an array of strings (`dtype='<U7'`): `"YYYY-MM"`,
- `np_prices` is initially an array of strings as well; we will cast it to `float` later.

Printing the arrays lets us visually confirm that the parsing step worked and that the data matches the original CSV file.

```python
np_dates = np.array(dates)
np_prices = np.array(prices, dtype=float)

print(np_dates[:10])
print(np_prices[:10])
```

```python
years = []
months = []

for d in np_dates:
    y_str, m_str = d.split("-")
    years.append(int(y_str))
    months.append(int(m_str))

np_years = np.array(years)
np_months = np.array(months)
```

## 5. Splitting `YYYY-MM` into `year` and `month`

The `Date` column is stored as a string in the format `YYYY-MM`.  
For many analyses it is more convenient to have **separate numeric columns**:

- `year` (e.g. 1998),
- `month` (1–12).

We loop over `np_dates`, split each string on the dash, and convert both parts to integers.  
Finally, we wrap the resulting Python lists into NumPy arrays `np_years` and `np_months`.

```python
np_prices = np.array(prices, dtype= float)

table = np.column_stack ((np_years, np_months, np_prices))
print(table.shape)
print(table[:5])
```

## 6. Building the final 2D NumPy table

Now we have three aligned 1D arrays of length *N*:

- `np_years`  – integer year values,
- `np_months` – integer month (1–12),
- `np_prices` – monthly gold prices as floats.

We combine them column‑wise using `np.column_stack` to create a single 2D array:

- each **row** corresponds to one month,
- each **column** corresponds to a feature: `[year, month, price]`.

We also inspect:

- `table.shape` – to confirm the number of rows and columns,
- the first few rows – to verify that the final representation looks reasonable.
