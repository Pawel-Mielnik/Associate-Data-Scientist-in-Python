```python
# Load the monthly gold price data from the CSV file.
# The file is stored in the data/ folder one level above this notebook.

import pandas as pd

df = pd.read_csv("../data/gold-price-analysis.csv")
```

```python
# === Basic statistics for the full dataset ===
# Goal: Get a quick overview of the entire time series.
# We calculate:
# - the minimum price,
# - the maximum price,
# - the mean (average) price,
# - the median price,
# - and the total number of monthly records.

min_price = df["Price"].min()
max_price = df["Price"].max()
avg_price = df["Price"].mean()
med_price = df["Price"].median()
total_records = len(df)
```

```python
# === Recent period (last 30 years) ===
# Idea: Instead of comparing 1830s to 2020s directly,
# we focus on a more realistic modern period.
# Here we filter the data to keep only the last ~30 years
# and compute the same statistics (min, max, mean, median, count)
# to see how recent prices compare to the full history.

# make sure Date is treated as a datetime column
df["Date"] = pd.to_datetime(df["Date"])

# keep only the last 30 years from the most recent date
last_date = df["Date"].max()
cutoff_date = last_date - pd.DateOffset(years=30)

recent_df = df[df["Date"] >= cutoff_date]

recent_min_price = recent_df["Price"].min()
recent_max_price = recent_df["Price"].max()
recent_avg_price = recent_df["Price"].mean()
recent_med_price = recent_df["Price"].median()
recent_total_records = len(recent_df)

recent_min_price, recent_max_price, recent_avg_price, recent_med_price, recent_total_records
```

```python
# === Types in action: int, float, str, bool ===
# In this section we explicitly use the basic Python types:
# - float: for numeric prices (min, max, mean, median),
# - int: for counts, e.g. number of records,
# - bool: for a simple condition, e.g. is the latest price higher than one year ago?,
# - str: to build a human-readable text summary of our findings.
# The goal is to connect the gold price example with Python's core data types.

# floats – numeric statistics for the full dataset
min_price = float(df["Price"].min())
max_price = float(df["Price"].max())
avg_price = float(df["Price"].mean())
med_price = float(df["Price"].median())

# int – counts
total_records = int(len(df))
recent_records = int(len(recent_df))

# bool – simple condition: is latest price higher than 12 months ago?
latest_price = float(df["Price"].iloc[-1])
price_12_months_ago = float(df["Price"].iloc[-13])  # 12 full months back
is_price_higher_than_last_year = latest_price > price_12_months_ago

# str – human-readable summary
summary = (
    "Gold price analysis summary\n"
    f"- Total monthly records: {total_records}\n"
    f"- Min price (full history): {min_price:.2f}\n"
    f"- Max price (full history): {max_price:.2f}\n"
    f"- Mean price (full history): {avg_price:.2f}\n"
    f"- Median price (full history): {med_price:.2f}\n"
    f"- Records in the last ~30 years: {recent_records}\n"
    f"- Is the latest price higher than 12 months ago? {is_price_higher_than_last_year}"
)

print(summary)
```
