# Exploratory Data Analysis of 1990s Netflix Movies

## Project goal
The goal of this project is to explore Netflix movie data and answer two questions about movies released in the 1990s:

1. What was the most frequent movie duration in the 1990s?
2. How many short action movies were released in the 1990s?

A short movie is defined as a movie with a duration of less than 90 minutes.

```python
import pandas as pd
import matplotlib.pyplot as plt

netflix_df = pd.read_csv("netflix_data.csv")
netflix_df.head()
```

## Data filtering

In this step, I filter the dataset to keep only movies released between 1990 and 1999.

I focus only on rows where:
- `type` is `"Movie"`
- `release_year` is between 1990 and 1999

```python
is_movie = netflix_df["type"] == "Movie"
is_1990s = (netflix_df["release_year"] >= 1990) & (netflix_df["release_year"] <= 1999)

movies_1990s = netflix_df[is_movie & is_1990s]
movies_1990s.head()
```

## Most frequent duration

To find the most common movie duration, I select the `duration` column from the filtered DataFrame and calculate its mode.

The mode is the value that appears most often in the dataset.

```python
most_common_duration = movies_1990s["duration"].mode()[0]
duration = most_common_duration

print(duration)
```

## Short action movies

In this step, I identify short action movies released in the 1990s.

I define a short movie as a movie with a duration of less than 90 minutes, and then filter the data to keep only action movies that meet this condition.

```python
is_action = movies_1990s["genre"] == "Action"
is_short = movies_1990s["duration"] < 90

short_action_movies_1990s = movies_1990s[is_action & is_short]
short_movie_count = short_action_movies_1990s.shape[0]

print(short_movie_count)
```
